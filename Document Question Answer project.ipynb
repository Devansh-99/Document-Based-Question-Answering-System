```python


from langchain.vectorstores import Chroma
from langchain import OpenAI,VectorDBQA
from langchain.document_loaders import UnstructuredFileLoader
from langchain.text_splitter import CharacterTextSplitter
import nltk
nltk.download('punkt')
```

```python
loader = UnstructuredFileLoader('/content/TCS-NQT Complete Aptitude.pdf')
documents = loader.load()
```

```python
text_splitter = CharacterTextSplitter(chunk_size=800, chunk_overlap=0)
texts = text_splitter.split_documents(documents)
```

```python
from google.colab import userdata
userdata.get('OPENAI_API_KEY')
```

```python
embeddings = OpenAIEmbeddings(openai_api_key = userdata.get('OPENAI_API_KEY'))
doc_search = Chroma.from_documents(texts,embeddings)
chain = VectorDBQA.from_chain_type(llm=OpenAI(), chain_type="stuff", vectorstore=doc_search)
query = input("what is the title")
chain.run(query)
```
